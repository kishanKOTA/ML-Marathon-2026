```python
import pandas as pd
import numpy as np
import tomllib
from PIL import Image
```

```python
from pathlib import Path

with open("../configs/default.toml", "rb") as f:
    data = tomllib.load(f)

# Resolve paths relative to the repo root (two levels up from notebook)
repo_root = Path("../../..").resolve()
split_path = repo_root / data['paths']['split']

print(split_path)
```

```python
df = pd.read_csv(str(split_path))
print(df['split'].value_counts())
```

```python
image_name = df['filename'][0]
image_path = repo_root / data['paths']['images'] / image_name
img = Image.open(image_path)
img_arr = np.asarray(img)

print(img_arr)
print(img.mode)
print(img.size)
display(img)
```
