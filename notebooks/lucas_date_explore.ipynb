```python
import pandas
import tomllib
from pathlib import Path
from PIL import Image
import numpy 
```

```python
with open("../configs/default.toml", "rb") as f:
    data = tomllib.load(f)

print(data)
print(data['paths']['split'])
split_path = data['paths']['split']

root = Path("../../..").resolve()
split_path = root / data['paths']['split']
print(split_path)

df = pandas.read_csv(split_path)

print(df)
print(df['split'].value_counts())
# df2 = pandas.read_csv('ML-Marathon-2026/data/processed/split.csv')
```

```python
image = df['filename'][0]
image_path = root / data['paths']['images'] / image
img = Image.open(image_path)
img_array = numpy.asarray(img)

print(img_array)
print(img.mode)
print(img.size)
img.show()
```
